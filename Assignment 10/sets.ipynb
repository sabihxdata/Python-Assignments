Find union, intersection, a - b, and symmetric difference.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

union = a.union(b)
intersection = a.intersection(b)
subtract = a - b

symmetric_diff = a.symmetric_difference(b)

print(union)
print(intersection)
print(subtract)
print(symmetric_diff)
```

Q27. Three-Set Analysis

Find students in all three, exactly Python and SQL (not AWS), and at least two.

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python.intersection(sql,aws)
python_sql_only = python.intersection(sql)-aws

python_sql = python.intersection(sql)
python_aws = python.intersection(aws)
sql_aws = sql.intersection(aws)

at_least_two = python_sql.union(python_aws,sql_aws)

print("All three courses:", all_three)
print("Python and SQL only:", python_sql_only)
print("At least two courses:", at_least_two)
```

Q28. Transaction Analysis

Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1.intersection(day2, day3)

only_day1 = day1 - day2 - day3
only_day2 = day2 - day1 - day3
only_day3 = day3 - day1 - day2

exactly_one = only_day1.union(only_day2, only_day3)

day1_day2 = day1.intersection(day2)
day1_day3 = day1.intersection(day3)
day2_day3 = day2.intersection(day3)

at_least_two = day1_day2.union(day1_day3, day2_day3)

print("IDs on all days:", all_days)
print("IDs on exactly one day:", exactly_one)
print("IDs on at least two days:", at_least_two)
```

Q29. Access Control

Find missing and extra permissions and test full authorization.

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions

extra = user_permissions - required_permissions

authorized = required_permissions.issubset(user_permissions)

print("missing permissions:", missing)
print("Extra permissions:", extra)
print("fully authorized:", authorized)
```

Q30. Customer Segmentation

Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website.intersection(mobile, store)

only_website = website - mobile - store
only_mobile = mobile - website - store
only_store = store - website - mobile

exactly_one = only_website.union(only_mobile, only_store)


website_mobile = (website.intersection(mobile)) - store
website_store = (website.intersection(store)) - mobile
mobile_store = (mobile.intersection(store)) - website

exactly_two = website_mobile.union(website_store, mobile_store)

print("Exactly one platform:", exactly_one)
print("Exactly Two platforms:", exactly_two)
print("All three platforms:", all_three)
```
