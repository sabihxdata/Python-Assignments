Q41. Student Performance System

Calculate each average, top student, tuple of names, and set of unique marks. 

```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

ali_avg = sum(students[0]["marks"]) / len(students[0]["marks"])
sara_avg = sum(students[1]["marks"]) / len(students[1]["marks"])
ahmed_avg = sum(students[2]["marks"]) / len(students[2]["marks"])

averages = {
    students[0]["name"]: ali_avg,
    students[1]["name"]: sara_avg,
    students[2]["name"]: ahmed_avg
}

top_student = max(averages, key=averages.get)

names = (
    students[0]["name"],
    students[1]["name"],
    students[2]["name"]
)

unique_marks = (
    set(students[0]["marks"])
    | set(students[1]["marks"])
    | set(students[2]["marks"])
)

print("Averages:", averages)
print("Top student:", top_student)
print("Names:", names)
print("Unique marks:", unique_marks)
```

Q45. Course Enrollment

Find students in exactly two, all three, and exactly one course; count unique students.

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python_sql = courses["Python"] & courses["SQL"]
python_aws = courses["Python"] & courses["AWS"]
sql_aws = courses["SQL"] & courses["AWS"]

all_three = (
    courses["Python"]
    & courses["SQL"]
    & courses["AWS"]
)

exactly_two = (
    (python_sql - courses["AWS"])
    | (python_aws - courses["SQL"])
    | (sql_aws - courses["Python"])
)

exactly_one = (
    (courses["Python"] - courses["SQL"] - courses["AWS"])
    | (courses["SQL"] - courses["Python"] - courses["AWS"])
    | (courses["AWS"] - courses["Python"] - courses["SQL"])
)

all_students = (
    courses["Python"]
    | courses["SQL"]
    | courses["AWS"]
)

print("Exactly two courses:", exactly_two)
print("All three courses:", all_three)
print("Exactly one course:", exactly_one)
print("Unique students:", len(all_students))
```

Q49. Product Recommendations

Find products bought by at least two users, Ali wishlist minus purchases, shared wishlist products, and all products.

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali_purchased = users["Ali"]["purchased"]
sara_purchased = users["Sara"]["purchased"]
ahmed_purchased = users["Ahmed"]["purchased"]

bought_by_two = (
    (ali_purchased & sara_purchased)
    | (ali_purchased & ahmed_purchased)
    | (sara_purchased & ahmed_purchased)
)

ali_recommendations = users["Ali"]["wishlist"] - users["Ali"]["purchased"]

shared_wishlist = (
    (users["Ali"]["wishlist"] & users["Sara"]["wishlist"])
    | (users["Ali"]["wishlist"] & users["Ahmed"]["wishlist"])
    | (users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"])
)

all_products = (
    ali_purchased
    | sara_purchased
    | ahmed_purchased
    | users["Ali"]["wishlist"]
    | users["Sara"]["wishlist"]
    | users["Ahmed"]["wishlist"]
)

print("Bought by at least two users:", bought_by_two)
print("Ali's wishlist minus purchases:", ali_recommendations)
print("Shared wishlist products:", shared_wishlist)
print("All products:", all_products)
```

Q50. Final Company Analysis

1. Payroll per department. 2. Company payroll. 3. Highest-paid employee. 4. Skills shared by all. 5. Engineering-only
skills. 6. Remaining budget per department. 7. Department with higher average salary. 8. Name-to-salary dictionary 9.All unique skills. 10. Is combined budget sufficient?

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]

engineering_payroll = (
    engineering["employees"][0][1]
    + engineering["employees"][1][1]
)

analytics_payroll = (
    analytics["employees"][0][1]
    + analytics["employees"][1][1]
)

company_payroll = engineering_payroll + analytics_payroll

salaries = {
    engineering["employees"][0][0]: engineering["employees"][0][1],
    engineering["employees"][1][0]: engineering["employees"][1][1],
    analytics["employees"][0][0]: analytics["employees"][0][1],
    analytics["employees"][1][0]: analytics["employees"][1][1]
}

highest_paid = max(salaries, key=salaries.get)

shared_skills = (
    engineering["employees"][0][2]
    & engineering["employees"][1][2]
    & analytics["employees"][0][2]
    & analytics["employees"][1][2]
)

engineering_skills = (
    engineering["employees"][0][2]
    | engineering["employees"][1][2]
)

analytics_skills = (
    analytics["employees"][0][2]
    | analytics["employees"][1][2]
)

engineering_only = engineering_skills - analytics_skills

engineering_remaining = engineering["budget"] - engineering_payroll
analytics_remaining = analytics["budget"] - analytics_payroll

engineering_average = engineering_payroll / len(engineering["employees"])
analytics_average = analytics_payroll / len(analytics["employees"])

higher_average = (
    "Engineering"
    if engineering_average > analytics_average
    else "Analytics"
)

all_skills = engineering_skills | analytics_skills

combined_budget = engineering["budget"] + analytics["budget"]
budget_sufficient = combined_budget >= company_payroll

print("Engineering payroll:", engineering_payroll)
print("Analytics payroll:", analytics_payroll)
print("Company payroll:", company_payroll)
print("Highest-paid employee:", highest_paid)
print("Skills shared by all:", shared_skills)
print("Engineering-only skills:", engineering_only)
print("Engineering remaining budget:", engineering_remaining)
print("Analytics remaining budget:", analytics_remaining)
print("Department with higher average salary:", higher_average)
print("Name-to-salary dictionary:", salaries)
print("All unique skills:", all_skills)
print("Combined budget sufficient:", budget_sufficient)
```
